```python
from ultralytics import RTDETR, YOLO
import albumentations as A
from pathlib import Path, PosixPath
```

```python
ds_yamls = [PosixPath('Dataset/Aircraft_Fuselage_DET2023/aircraft_fuselage_yolo/2026-01-30_5-Fold_Cross-val/split_1/split_1_dataset.yaml'),
 PosixPath('Dataset/Aircraft_Fuselage_DET2023/aircraft_fuselage_yolo/2026-01-30_5-Fold_Cross-val/split_2/split_2_dataset.yaml'),
 PosixPath('Dataset/Aircraft_Fuselage_DET2023/aircraft_fuselage_yolo/2026-01-30_5-Fold_Cross-val/split_3/split_3_dataset.yaml'),
 PosixPath('Dataset/Aircraft_Fuselage_DET2023/aircraft_fuselage_yolo/2026-01-30_5-Fold_Cross-val/split_4/split_4_dataset.yaml'),
 PosixPath('Dataset/Aircraft_Fuselage_DET2023/aircraft_fuselage_yolo/2026-01-30_5-Fold_Cross-val/split_5/split_5_dataset.yaml')]
```

```python

results = {}

# Define your additional arguments here
batch = 8
project = "kfold_aug_rtdetr-x-2"
epochs = 50
model_name = "rtdetr-x.pt"

# only one for now
for k, dataset_yaml in enumerate(ds_yamls[:1]):
    A.MotionBlur(p=0.2),
    A.Rotate(limit=20, p=0.3),
    model = RTDETR(model_name)
    results[k] = model.train(
        data=dataset_yaml,
        epochs=epochs,
        batch=batch,
        project=project,
        name=f"fold_{k + 1}",
        imgsz=640,
        patience=25,
    )
```
